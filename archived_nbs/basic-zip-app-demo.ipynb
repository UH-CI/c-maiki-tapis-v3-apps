## Systems

In this section we create a Tapis system, one for running on a VM host using FORK and one for running on an HPC type host using BATCH.

### Create a system for the VM host

Simply run these to create a Tapis system

```python
# IDs in Tapis must be "globally" unique, not just to the user
user_id = vm_username
system_id_vm = "test-app-" + user_id
```

```python
# Create the system definition 
# PASSWORD FOR AUTH EXAMPLE
exec_system_vm = {
  "id": system_id_vm,
  "description": "Test vm system using password for auth",
  "systemType": "LINUX", 
  "host": host,
  "effectiveUserId":"exouser",
  "defaultAuthnMethod": "PASSWORD",
  "rootDir": "/",
  "canExec": True,
  "jobRuntimes": [ { "runtimeType": "ZIP" } ],
  "jobWorkingDir": "HOST_EVAL($HOME)/sharetest/workdir"
}

# Use the client to create the system in Tapis
print("****************************************************")
print("Create system: " + system_id_vm)
print("****************************************************")
client.systems.createSystem(**exec_system_vm)
```

```python
# List all systems available to you
print("****************************************************")
print("List all systems")
print("****************************************************")
client.systems.getSystems()
```

```python
# Get details for the system you created
print("****************************************************")
print("Fetch system: " + system_id_vm)
print("****************************************************")
client.systems.getSystem(systemId=system_id_vm)
```

### Register Credentials for the VM system

After creating the system, you will need to register credentials for your username. These will be used by Tapis to
access the host. Various authentication methods can be used to access a system, such as PASSWORD and PKI_KEYS. For the
VM a password is used.

```python
# Register credentials using PASSWORD
# Credentials should align with the system to be used, not necessarily the TACC credentials
client.systems.createUserCredential(systemId=system_id_vm, userName=vm_username, password=vm_password)
```

### App definition

```python
# Create appplication definition
# Jobs to be run from ZIP archive files need FORK jobType
# containerImage should point to ZIP containing necessary files
# At minimum, a script file named tapisjob_app.sh to drive the app

app_id = "test-zip-app-" + user_id
app_def= {
    "id": app_id,
    "version": "0.1",
    "description": "Test running a zip application.",
    "jobType": "FORK",
    "runtime": "ZIP",
    # This should be the path to your zipped application (airrflow and tapisjob_app.sh)
    # The path likely starts with tapis://test-app-exouser/home/exouser, depending on how you set up
    # In the example below, I have a folder called "apps" on my VM and the app is packaged in test.tar.gz
    # Your case may be something like tapis://test-app-exouser/home/exouser/apps/airrflow-app.tar.gz
    # That is if you create a folder called apps and put your app there
    "containerImage": "tapis://test-app-exouser/home/exouser/apps/test.tar.gz",
    "jobAttributes": {
        "parameterSet": {
            "archiveFilter": {
                "includeLaunchFiles": False
            }
        },
        # Shouldn't have to worry about these unless you get specific errors about memory, cores, nodes
        "memoryMB": 3000,
        "nodeCount": 1,
        "coresPerNode": 1,
        # This is for adjusting how long you allow the job to run (in minutes)
        "maxMinutes": 720
    }
}
```

### Create App

```python
# To create the app
client.apps.createAppVersion(**app_def)
```

### Update app

```python
# To update the app, appVersion must match the version you want to update
client.apps.patchApp(appId=app_id, appVersion='0.1', **app_def)
```

```python
client.apps.getAppLatestVersion(appId=app_id)
```

```python
# Submit job, pass any necessary arguments
# More information about parameterSet: https://tapis.readthedocs.io/en/latest/technical/jobs.html#parameterset
pa= {
    "parameterSet": {
    "appArgs": [
#         {"arg": "--text 'I am happy today'"}

        ]
    }
}

# Submit a job
job_response_vm=client.jobs.submitJob(name='Test job',description='Testing',appId=app_id,appVersion='0.1',execSystemId=system_id_vm, **pa)
```

```python
# Get Job submission response
print("****************************************************")
print("Job Submitted: " + app_id)
print("****************************************************")
print(job_response_vm)
```

```python
# Get job uuid from the job submission response
print("****************************************************")
job_uuid_vm=job_response_vm.uuid
print("Job UUID: " + job_uuid_vm)
print("****************************************************")
```
